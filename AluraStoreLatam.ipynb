### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
ingresos = {
    "Tienda 1": tienda['Precio'].sum(),
    "Tienda 2": tienda2['Precio'].sum(),
    "Tienda 3": tienda3['Precio'].sum(),
    "Tienda 4": tienda4['Precio'].sum()
}

for tienda, ingreso in ingresos.items():
    print(f"{tienda}: ${ingreso:,.2f}")
```

# 2. Ventas por categoría

```python
categorias_tienda1 = tienda.groupby('Categoría del Producto').size().sort_values(ascending=False)
categorias_tienda2 = tienda2.groupby('Categoría del Producto').size().sort_values(ascending=False)
categorias_tienda3 = tienda3.groupby('Categoría del Producto').size().sort_values(ascending=False)
categorias_tienda4 = tienda4.groupby('Categoría del Producto').size().sort_values(ascending=False)

print("Tienda 1")
print(categorias_tienda1)

print("\nTienda 2")
print(categorias_tienda2)

print("\nTienda 3")
print(categorias_tienda3)

print("\nTienda 4")
print(categorias_tienda4)
```

# 3. Calificación promedio de la tienda


```python
calificacion_tienda1 = tienda['Calificación'].mean()
calificacion_tienda2 = tienda2['Calificación'].mean()
calificacion_tienda3 = tienda3['Calificación'].mean()
calificacion_tienda4 = tienda4['Calificación'].mean()

print("Calificación promedio Tienda 1:", calificacion_tienda1)
print("Calificación promedio Tienda 2:", calificacion_tienda2)
print("Calificación promedio Tienda 3:", calificacion_tienda3)
print("Calificación promedio Tienda 4:", calificacion_tienda4)
```

# 4. Productos más y menos vendidos

```python
print("Tienda 1")
ventas_tienda1 = tienda['Producto'].value_counts()
print("Producto más vendido:")
print(ventas_tienda1.head(1))
print("Producto menos vendido:")
print(ventas_tienda1.tail(1))


print("\nTienda 2")
ventas_tienda2 = tienda2['Producto'].value_counts()
print("Producto más vendido:")
print(ventas_tienda2.head(1))
print("Producto menos vendido:")
print(ventas_tienda2.tail(1))


print("\nTienda 3")
ventas_tienda3 = tienda3['Producto'].value_counts()
print("Producto más vendido:")
print(ventas_tienda3.head(1))
print("Producto menos vendido:")
print(ventas_tienda3.tail(1))


print("\nTienda 4")
ventas_tienda4 = tienda4['Producto'].value_counts()
print("Producto más vendido:")
print(ventas_tienda4.head(1))
print("Producto menos vendido:")
print(ventas_tienda4.tail(1))
```

# 5. Envío promedio por tienda

```python
# Costo de envío promedio por tienda

envio_tienda1 = tienda['Costo de envío'].mean()
envio_tienda2 = tienda2['Costo de envío'].mean()
envio_tienda3 = tienda3['Costo de envío'].mean()
envio_tienda4 = tienda4['Costo de envío'].mean()

print("Costo de envío promedio Tienda 1:", envio_tienda1)
print("Costo de envío promedio Tienda 2:", envio_tienda2)
print("Costo de envío promedio Tienda 3:", envio_tienda3)
print("Costo de envío promedio Tienda 4:", envio_tienda4)
```

#Grafico

```python
import matplotlib.pyplot as plt
```

```python
ingresos = [
    tienda['Precio'].sum(),
    tienda2['Precio'].sum(),
    tienda3['Precio'].sum(),
    tienda4['Precio'].sum()
]

nombres_tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']

plt.bar(nombres_tiendas, ingresos)
plt.title("Ingresos totales por tienda")
plt.xlabel("Tiendas")
plt.ylabel("Ingresos")
plt.show()
```

```python
categorias = tienda['Categoría del Producto'].value_counts()

plt.pie(categorias, labels=categorias.index, autopct='%1.1f%%')
plt.title("Distribución de ventas por categoría - Tienda 1")
plt.show()
```

```python
plt.scatter(tienda['Precio'], tienda['Calificación'])
plt.title("Relación entre precio y calificación - Tienda 1")
plt.xlabel("Precio")
plt.ylabel("Calificación")
plt.show()
```
